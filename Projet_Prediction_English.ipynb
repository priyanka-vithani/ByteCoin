2. **Analysis of attribute distributions**:

    - Examine the distribution of each attribute using appropriate visualizations such as histograms and boxplots. Discuss the insights obtained, including the presence of outliers.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Histograms for all numeric columns
dataset_att = dataset.drop(['diabetes'], axis=1)


# Separate numeric and categorical columns
numeric_cols = dataset_att.select_dtypes(include=['float64', 'int64']).columns
categorical_cols = dataset.select_dtypes(include=['object']).columns

# Histograms for all numeric columns
dataset_att[numeric_cols].hist(figsize=(10, 8))
plt.show()

# Boxplots for numeric columns
plt.figure(figsize=(12, 6))
sns.boxplot(data=dataset_att[numeric_cols])
plt.xticks(rotation=90)
plt.show()

# Bar plots for categorical columns
for col in categorical_cols:
    plt.figure(figsize=(6, 4))
    sns.countplot(x=dataset_att[col])
    plt.title(f'Distribution of {col}')
    plt.show()

```
